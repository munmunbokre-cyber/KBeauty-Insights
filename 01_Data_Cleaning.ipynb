```python
df = df_raw.copy()

df = df[
    df["product_name"] != "제품이름"
].copy()

print("Extra header row removed!")

print(df.shape)
```

```python
df = df.rename(
    columns={"Unnamed: 0": "source_id"}
)

print(df.columns.tolist())
```

```python
df["rating"] = pd.to_numeric(
    df["rating"],
    errors="coerce"
)

print(df["rating"].value_counts(dropna=False))
```

```python
df = df[
    df["rating"].between(1, 5)
].copy()
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna(subset=["review"])
```

```python
df["skin_type"] = df["skin_type"].fillna(
    "Unknown"
)
```

```python
print(
    "Duplicate Records:",
    df.duplicated(
        subset=[
            "product_name",
            "rating",
            "skin_type",
            "review"
        ]
    ).sum()
)
```

```python
def clean_korean_text(text):

    text = str(text)

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Remove unnecessary spaces
    text = text.replace("\xa0", " ")

    # Remove extra whitespace
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
df["cleaned_review"] = df["review"].apply(
    clean_korean_text
)
```

```python
display(
    df[["review", "cleaned_review"]].head(10)
)
```

```python
def assign_sentiment(rating):

    if rating <= 2:
        return "Negative"

    elif rating == 3:
        return "Neutral"

    else:
        return "Positive"
```

```python
df["sentiment"] = df["rating"].apply(
    assign_sentiment
)

display(
    df[["rating", "sentiment"]].head(10)
)
```

```python
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nRating Distribution:")
print(df["rating"].value_counts())

print("\nSentiment Distribution:")
print(df["sentiment"].value_counts())

display(df.head())
```

```python
df.to_csv(
    "korean_beauty_reviews_cleaned.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Cleaned dataset exported successfully!")
```
